```python
from anytree import Node, RenderTree
from tkinter import Tk, ttk

# 1. 建立 anytree 樹
root = Node("Root")
child1 = Node("Child1", parent=root)
child2 = Node("Child2", parent=root)
subchild1 = Node("SubChild1", parent=child1)
subchild2 = Node("SubChild2", parent=child2)

# 2. 建立 tkinter 視窗
window = Tk()
window.title("AnyTree TreeView Example")
window.geometry("300x300")

# 3. 建立 Treeview 元件
tree = ttk.Treeview(window)
tree.pack(expand=True, fill="both")

# 4. 遞迴加入節點
def insert_node(parent_tk_id, anytree_node):
    # 加入此節點
    node_id = tree.insert(parent_tk_id, "end", text=anytree_node.name)

    # 對所有子節點遞迴
    for child in anytree_node.children:
        insert_node(node_id, child)

# 5. 開始加入根節點
insert_node("", root)

# 6. 執行 tkinter 主迴圈
window.mainloop()
```
